````python
import sys
sys.path.insert(0, '/content')

from trust_layer.record import PatientRecord
from trust_layer.validators import (
    validate_completeness, validate_temporal,
    validate_identity_consistency, validate_provenance,
    CrossRecordValidator,
)
from trust_layer.scoring import compute_trust_score
from trust_layer.router import route_decision, route_decision_hard


def test_full_pipeline_clean_record():
    cv = CrossRecordValidator()
    rec = PatientRecord(
        emirates_id="784-1985-1234567-1",
        given_name="Ahmed", family_name="Al-Mansoori",
        date_of_birth="1985-03-15", nationality="UAE",
        source_facility="Cleveland Clinic Abu Dhabi",
        registration_date="2024-01-10",
        canonical_id="PATIENT_001",
    )
    dims = {
        "completeness": validate_completeness(rec),
        "temporal": validate_temporal(rec),
        "identity": validate_identity_consistency(rec),
        "provenance": validate_provenance(rec),
        "cross_record": cv.validate(rec),
    }
    score = compute_trust_score(**dims)
    decision = route_decision(score)
    assert decision == "AUTO_LINK", f"Expected AUTO_LINK, got {decision}"
    print(f"Clean record: score={score}, decision={decision}")


def test_full_pipeline_collision():
    cv = CrossRecordValidator()
    patient_a = PatientRecord(
        emirates_id="784-1985-1234567-1",
        given_name="Ahmed", family_name="Al-Mansoori",
        date_of_birth="1985-03-15", nationality="UAE",
        source_facility="Cleveland Clinic Abu Dhabi",
        registration_date="2024-01-10",
        canonical_id="PATIENT_A",
    )
    cv.add_record(patient_a)

    colliding = PatientRecord(
        emirates_id="784-1985-1234567-1",
        given_name="Fatima", family_name="Al-Zahra",
        date_of_birth="1992-01-01", nationality="UAE",
        source_facility="Cleveland Clinic Abu Dhabi",
        registration_date="2024-06-01",
        canonical_id="PATIENT_B",
    )
    dims = {
        "completeness": validate_completeness(colliding),
        "temporal": validate_temporal(colliding),
        "identity": validate_identity_consistency(colliding),
        "provenance": validate_provenance(colliding),
        "cross_record": cv.validate(colliding),
    }
    score = compute_trust_score(**dims)
    decision = route_decision_hard(score, dims["cross_record"])
    assert decision == "QUARANTINE", f"Expected QUARANTINE, got {decision}"
    assert dims["cross_record"] == 0.0
    print(f"Collision: score={score}, decision={decision} (hard override)")


test_full_pipeline_clean_record()
test_full_pipeline_collision()
print("Integration tests passed.")
````

````python
# Update trust_layer/audit.py to add hash-chaining
with open('/content/trust_layer/audit.py', 'w') as f:
    f.write('''import json
import os
import hashlib
from datetime import datetime


AUDIT_LOG_PATH = "audit_log.jsonl"


def _hash_entry(entry, previous_hash):
    payload = json.dumps(entry, sort_keys=True) + "|" + previous_hash
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


def _last_hash(path):
    if not os.path.exists(path):
        return "GENESIS"
    with open(path, "rb") as f:
        lines = f.read().splitlines()
        if not lines:
            return "GENESIS"
        try:
            last = json.loads(lines[-1].decode("utf-8"))
            return last.get("hash", "GENESIS")
        except Exception:
            return "GENESIS"


def log_decision(record, dimensions, score, decision, log_path=None):
    path = log_path or AUDIT_LOG_PATH
    prev_hash = _last_hash(path)

    entry = {
        "timestamp": datetime.utcnow().isoformat() + "Z",
        "canonical_id": record.canonical_id,
        "identifier_present": bool(record.emirates_id),
        "source_facility": record.source_facility,
        "dimensions": dimensions,
        "trust_score": score,
        "decision": decision,
    }
    entry["hash"] = _hash_entry(entry, prev_hash)

    with open(path, "a") as f:
        f.write(json.dumps(entry) + "\\n")
    return entry


def verify_chain(log_path=None):
    path = log_path or AUDIT_LOG_PATH
    if not os.path.exists(path):
        return True, "no log"
    prev = "GENESIS"
    with open(path, "rb") as f:
        for i, line in enumerate(f):
            try:
                entry = json.loads(line.decode("utf-8"))
            except Exception:
                return False, f"malformed line {i}"
            stored_hash = entry.pop("hash", None)
            computed = _hash_entry(entry, prev)
            if computed != stored_hash:
                return False, f"chain broken at line {i}"
            prev = stored_hash
    return True, "ok"
''')

print("audit.py updated with hash-chaining")
````

````python
import sys, random
sys.path.insert(0, '/content')

import recordlinkage
from recordlinkage.datasets import load_febrl1
from trust_layer.record import PatientRecord
from trust_layer.validators import CrossRecordValidator

random.seed(42)

# Load Febrl1 and split originals from duplicates
df = load_febrl1().reset_index()
df['orig_num'] = df['rec_id'].str.extract(r'rec-(\d+)-')
df['is_dup'] = df['rec_id'].str.contains('-dup-')
df_a = df[~df['is_dup']].copy()


def febrl_to_record(row, canonical_id):
    dob_raw = str(row.get('date_of_birth', '') or '')
    if len(dob_raw) == 8 and dob_raw.isdigit():
        dob = f"{dob_raw[:4]}-{dob_raw[4:6]}-{dob_raw[6:]}"
    else:
        dob = dob_raw
    return PatientRecord(
        emirates_id=str(row.get('soc_sec_id', '') or '') or None,
        given_name=str(row.get('given_name', '') or ''),
        family_name=str(row.get('surname', '') or ''),
        date_of_birth=dob,
        nationality='AU',
        source_facility='Royal Melbourne Hospital',
        registration_date='2024-01-10',
        canonical_id=canonical_id,
    )


febrl_records = [febrl_to_record(row, f"FEBRL_{i}") for i, row in df_a.iterrows()]

# Build reference MPI
cv = CrossRecordValidator()
for rec in febrl_records:
    cv.add_record(rec)


def inject_easy(target, donor):
    """Same identifier, different name and DOB."""
    return PatientRecord(
        emirates_id=donor.emirates_id,
        given_name=target.given_name or "X",
        family_name=target.family_name or "Y",
        date_of_birth=target.date_of_birth or "1950-01-01",
        source_facility=target.source_facility,
        canonical_id=target.canonical_id,
    )


def inject_medium(target, donor):
    """Near-match identifier: flip one digit."""
    eid = donor.emirates_id or "1234567"
    if len(eid) > 0:
        # Flip the last digit
        last = str((int(eid[-1]) + 1) % 10) if eid[-1].isdigit() else "0"
        eid = eid[:-1] + last
    return PatientRecord(
        emirates_id=eid,
        given_name=target.given_name or "X",
        family_name=target.family_name or "Y",
        date_of_birth=target.date_of_birth or "1950-01-01",
        source_facility=target.source_facility,
        canonical_id=target.canonical_id,
    )


def inject_hard(target, donor):
    """Same name, same DOB, no identifier match — spoof pattern."""
    return PatientRecord(
        emirates_id="9999999",
        given_name=target.given_name or "X",
        family_name=target.family_name or "Y",
        date_of_birth=target.date_of_birth or "1950-01-01",
        source_facility=target.source_facility,
        canonical_id=target.canonical_id,
    )


def inject_adversarial(target, donor):
    """Same name, same DOB, fresh identifier — A3 attack."""
    return PatientRecord(
        emirates_id="0000000",
        given_name=target.given_name or "X",
        family_name=target.family_name or "Y",
        date_of_birth=target.date_of_birth or "1950-01-01",
        source_facility=target.source_facility,
        canonical_id=target.canonical_id,
    )


# Run the ladder
N = 100
attack_types = [
    ("easy (identical ID)", inject_easy),
    ("medium (ID one digit off)", inject_medium),
    ("hard (name+DOB, no ID match)", inject_hard),
    ("adversarial (A3 pattern)", inject_adversarial),
]

print(f"Febrl collision detection ladder (n={N} injections each)")
print("=" * 62)

for label, inject_fn in attack_types:
    caught = 0
    for i in range(N):
        target_idx = random.randint(0, len(febrl_records) - 1)
        donor_idx = target_idx
        while donor_idx == target_idx:
            donor_idx = random.randint(0, len(febrl_records) - 1)

        colliding = inject_fn(febrl_records[target_idx], febrl_records[donor_idx])
        score = cv.validate(colliding)
        if score == 0.0:
            caught += 1

    print(f"{label}: {caught}/{N} = {100*caught/N:.1f}%")
````

````python
# Add auth to app.py
app_code = '''
import os
from fastapi import FastAPI, HTTPException, Header
from pydantic import BaseModel
from typing import Optional

from trust_layer.record import PatientRecord
from trust_layer.validators import (
    validate_completeness, validate_temporal,
    validate_identity_consistency, validate_provenance,
    CrossRecordValidator,
)
from trust_layer.scoring import compute_trust_score
from trust_layer.router import route_decision_hard


API_KEY = os.environ.get("TRUST_LAYER_API_KEY", "dev-key-change-in-production")
CONFIG_VERSION = "v0.2.0"


app = FastAPI(
    title="Identity Trust Assessment API",
    description="Five-dimension trust assessment for patient identity resolution in HIEs.",
    version=CONFIG_VERSION,
)

cross_validator = CrossRecordValidator()


class PatientInput(BaseModel):
    emirates_id: Optional[str] = None
    given_name: str = ""
    family_name: str = ""
    date_of_birth: str = ""
    nationality: str = ""
    source_facility: str = ""
    registration_date: str = ""
    canonical_id: str = ""


def check_auth(x_api_key: Optional[str]):
    if x_api_key != API_KEY:
        raise HTTPException(status_code=401, detail="Invalid or missing API key")


@app.get("/health")
def health():
    return {"status": "ok", "config_version": CONFIG_VERSION}


@app.post("/assess")
def assess(payload: PatientInput, x_api_key: Optional[str] = Header(None)):
    check_auth(x_api_key)
    record = PatientRecord(**payload.dict())

    dims = {
        "completeness": validate_completeness(record),
        "temporal": validate_temporal(record),
        "identity": validate_identity_consistency(record),
        "provenance": validate_provenance(record),
        "cross_record": cross_validator.validate(record),
    }
    score = compute_trust_score(**dims)
    decision = route_decision_hard(score, dims["cross_record"])

    return {
        "trust_score": score,
        "decision": decision,
        "dimensions": dims,
        "config_version": CONFIG_VERSION,
    }


@app.post("/register")
def register(payload: PatientInput, x_api_key: Optional[str] = Header(None)):
    check_auth(x_api_key)
    record = PatientRecord(**payload.dict())
    cross_validator.add_record(record)
    return {"registered": record.canonical_id, "config_version": CONFIG_VERSION}
'''

with open('/content/app.py', 'w') as f:
    f.write(app_code)

# Update audit.py to stamp config version
audit_patch = '''
import json
import os
import hashlib
from datetime import datetime


AUDIT_LOG_PATH = "audit_log.jsonl"
CONFIG_VERSION = "v0.2.0"
THRESHOLDS = {"low": 0.952, "medium": 0.571}


def _hash_entry(entry, previous_hash):
    payload = json.dumps(entry, sort_keys=True) + "|" + previous_hash
    return hashlib.sha256(payload.encode("utf-8")).hexdigest()


def _last_hash(path):
    if not os.path.exists(path):
        return "GENESIS"
    with open(path, "rb") as f:
        lines = f.read().splitlines()
        if not lines:
            return "GENESIS"
        try:
            last = json.loads(lines[-1].decode("utf-8"))
            return last.get("hash", "GENESIS")
        except Exception:
            return "GENESIS"


def log_decision(record, dimensions, score, decision, log_path=None):
    path = log_path or AUDIT_LOG_PATH
    prev_hash = _last_hash(path)

    entry = {
        "timestamp": datetime.utcnow().isoformat() + "Z",
        "config_version": CONFIG_VERSION,
        "thresholds": THRESHOLDS,
        "canonical_id": record.canonical_id,
        "identifier_present": bool(record.emirates_id),
        "source_facility": record.source_facility,
        "dimensions": dimensions,
        "trust_score": score,
        "decision": decision,
    }
    entry["hash"] = _hash_entry(entry, prev_hash)

    with open(path, "a") as f:
        f.write(json.dumps(entry) + "\\n")
    return entry


def verify_chain(log_path=None):
    path = log_path or AUDIT_LOG_PATH
    if not os.path.exists(path):
        return True, "no log"
    prev = "GENESIS"
    with open(path, "rb") as f:
        for i, line in enumerate(f):
            try:
                entry = json.loads(line.decode("utf-8"))
            except Exception:
                return False, f"malformed line {i}"
            stored_hash = entry.pop("hash", None)
            computed = _hash_entry(entry, prev)
            if computed != stored_hash:
                return False, f"chain broken at line {i}"
            prev = stored_hash
    return True, "ok"
'''

with open('/content/trust_layer/audit.py', 'w') as f:
    f.write(audit_patch)

print("app.py secured with API key + config version")
print("audit.py stamped with config_version and thresholds")
````
